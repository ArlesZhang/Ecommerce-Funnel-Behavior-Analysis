````python
```图类型

折线图 plot
柱状图 bar   
条线图 barh  函数签名是: plt.barh(y, width, height=0.8, left=None, *, align='center', **kwargs)
饼状图 pie
散点图 scatter
箱线图 boxplot
多个图表
组合图

绘制图表的 5 要素:
1.import 工具箱 2.创建图表(设置大小) 3.要绘图的数据 4.开始绘制plt.plot() 5.show 图表
```
````

````python
# 绘制-折线图 (大部分的参数都是通用的) 
import matplotlib.pyplot as plt

plt.rcdefaults()     # 恢复 Matplotlib 默认配置
plt.rcParams['font.sans-serif'] = ['Noto Sans CJK SC']  # 设置简体中文字体
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题 

# 要绘图的数据
month = ['1月','2月','3月','4月']   # 该库对中文的月份显示可能存在问题 需要Import rcParams 处理
sales = [100,150,80,130]

plt.figure(figsize=(10,5)) # 创建图表,设置大小
plt.plot(month,sales,     # 绘制折线图 (x,y) --表示x轴 y轴
         label='成都',
         color='orange',
         linewidth=2,
         linestyle='--',
         marker='o'
        )

plt.legend(loc='upper left')  # 添加图例的位置
plt.title('2026年质检量',color='red',fontsize=20)  # 添加标题

plt.xlabel('月份',fontsize=10)  # 添加坐标轴的标签 
plt.ylabel('销售额(万元)',fontsize=10) 
plt.ylim(0,160)  # 设置坐标轴的范围
plt.yticks(rotation=0,fontsize=10)  # 设置x y轴的刻度 ratation=字体旋转的角度

# 在每个数据点上显示数值 
for x,y in zip(month,sales):
    plt.text(x,y+1,str(y),ha='center',va='bottom',fontsize=10)  # ha=Horizontal Alignment 水平对齐 ; va=Vertical Alignment 垂直对齐

# 添加网格线 
# plt.grid(axis='y')  
plt.grid(True,alpha=0.1,color='blue',linestyle='--')  # 沿x y 轴的网格线, alpha 是设置透明度

# 显示图表
plt.show() 
````

````python
# 绘制-柱状图 (大部分的参数都是通用的) 
import matplotlib.pyplot as plt

plt.rcdefaults()     # 恢复 Matplotlib 默认配置
plt.rcParams['font.sans-serif'] = ['Noto Sans CJK SC']  # 设置简体中文字体
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题 

# 要绘图的数据
subjects = ['电销一组','电销二组','电销三组','电销四组']
scores = [85,92,78,88]

plt.figure(figsize=(10,5)) # 创建图表,设置大小

# 绘制-柱状图
plt.bar(subjects,scores,
        label='成都',
        color='orange',
        width=0.4
)

plt.legend(loc='upper right')  # 添加图例的位置
plt.title('2026年电销质检得分',color='red',fontsize=15)  # 添加标题

plt.xlabel('组别',fontsize=10)  # 添加坐标轴的标签 
plt.ylabel('分数',fontsize=10) 
plt.ylim(0,100)  # 设置坐标轴的范围
plt.yticks(rotation=0,fontsize=10)  # 设置x y轴的刻度 ratation=字体旋转的角度

# 在每个数据点上显示数值 
for x,y in zip(subjects,scores):
    plt.text(x,y+1,str(y),ha='center',va='bottom',fontsize=10)  # ha=Horizontal Alignment 水平对齐 ; va=Vertical Alignment 垂直对齐

# 添加网格线 
plt.grid(axis='y',alpha=0.1)  
# plt.grid(True,alpha=0.1,color='blue',linestyle='--')  # 沿x y 轴的网格线, alpha 是设置透明度

# 自动排版优化 
plt.tight_layout()
# 显示图表
plt.show() 
````

````python
# 绘制-条形图 (大部分的参数都是通用的) 
import matplotlib.pyplot as plt

plt.rcdefaults()     # 恢复 Matplotlib 默认配置
plt.rcParams['font.sans-serif'] = ['Noto Sans CJK SC']  # 设置简体中文字体
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题 

# 要绘图的数据
subjects = ['电销一组','电销二组','电销三组','电销四组']
scores = [85,92,78,88]

plt.figure(figsize=(10,5)) # 创建图表,设置大小

# 绘制-条形图
plt.barh(subjects,scores,
        label='成都',
        color='orange',
        height=0.6
)

plt.legend(loc='upper right')  # 添加图例的位置
plt.title('2026年电销质检得分',color='red',fontsize=15)  # 添加标题

plt.ylabel('组别',fontsize=10)  # 添加坐标轴的标签 
plt.xlabel('分数',fontsize=10) 
plt.xlim(0,100)  # 设置坐标轴的范围
plt.xticks(rotation=0,fontsize=10)  # 设置x y轴的刻度 ratation=字体旋转的角度

# 在每个数据点上显示数值 
for x,y in zip(subjects,scores):
    plt.text(y+1,x,str(y),ha='center',va='bottom',fontsize=10)  # ha=Horizontal Alignment 水平对齐 ; va=Vertical Alignment 垂直对齐

# 添加网格线 
plt.grid(axis='x',alpha=0.1)  
# plt.grid(True,alpha=0.1,color='blue',linestyle='--')  # 沿x y 轴的网格线, alpha 是设置透明度

# 自动排版优化 
plt.tight_layout()
# 显示图表
plt.show() 
````
